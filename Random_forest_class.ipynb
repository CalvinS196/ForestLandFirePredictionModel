```python
import pandas as pd
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
```

```python
modis = pd.read_csv('modis_2024_Indonesia.csv', sep=',')

modis.columns = modis.columns.str.strip()
kolom_teks = modis.select_dtypes(include='object').columns
modis[kolom_teks] = modis[kolom_teks].apply(lambda kolom: kolom.str.strip())

modis.head()
```

```python
#aman missing value
modis.isna().sum()
```

```python
# Distribusi tingkat FRP sebelum SMOTE
modis['tingkat_frp'] = pd.qcut(
    modis['frp'],
    q=3,
    labels=['Rendah', 'Sedang', 'Tinggi']
)

target = 'tingkat_frp'
distribusi_target = (
    modis[target]
    .value_counts()
    .rename_axis(target)
    .reset_index(name='jumlah')
)
distribusi_target['persentase'] = (
    distribusi_target['jumlah']
    / distribusi_target['jumlah'].sum()
    * 100
).round(2)

distribusi_target
```

```python
# Batas kategori FRP berdasarkan kuantil data
batas_kategori_frp = modis['frp'].quantile([0.33, 0.66])
batas_kategori_frp
```

```python
# Menyeimbangkan target tingkat_frp dengan SMOTE
# FRP dikeluarkan dari fitur karena digunakan untuk membentuk target
target = 'tingkat_frp'
X = modis.drop(columns=[target, 'frp'])
y = modis[target]

# SMOTE membutuhkan fitur numerik, sehingga fitur kategorikal diubah menjadi dummy
kolom_kategorikal_model = X.select_dtypes(include='object').columns
X_encoded = pd.get_dummies(X, columns=kolom_kategorikal_model, dtype=float)

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

distribusi_sebelum = y_train.value_counts().sort_index()

smote = SMOTE(random_state=42)
X_train, y_train = smote.fit_resample(X_train, y_train)


```

```python
# Distribusi tingkat FRP sesudah SMOTE
tingkat_frp_sesudah_smote = (
    y_train.value_counts()
    .rename_axis(target)
    .reset_index(name='jumlah')
)

tingkat_frp_sesudah_smote
```

Cleaning

```python
# Random Forest untuk klasifikasi tingkat FRP
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
)

model_rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
)
model_rf.fit(X_train, y_train)

prediksi_rf = model_rf.predict(X_test)

akurasi_rf = accuracy_score(y_test, prediksi_rf)
print(f'Akurasi: {akurasi_rf:.4f}')

print('\nClassification report:')
laporan_rf = pd.DataFrame(
    classification_report(y_test, prediksi_rf, output_dict=True)
).T.round(4)
print(laporan_rf)

print('\nConfusion matrix:')
matriks_kebingungan_rf = pd.DataFrame(
    confusion_matrix(y_test, prediksi_rf, labels=model_rf.classes_),
    index=[f'Aktual {kelas}' for kelas in model_rf.classes_],
    columns=[f'Prediksi {kelas}' for kelas in model_rf.classes_],
)
print(confusion_matrix(y_test, prediksi_rf))

```
